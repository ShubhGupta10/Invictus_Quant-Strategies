```python
import talib
import numpy as np
import pandas as pd
import math
import yfinance as yf

from backtesting import Backtest, Strategy
from backtesting.lib import crossover
from backtesting.test import GOOG
```

```python
cl = yf.Ticker("CL=F")
data = cl.history("max")
```

```python
class MA_MACD(Strategy):
    ma_window = 32
    fastperiod = 12
    slowperiod = 26
    signalperiod = 9
    
    def init(self):
        # Indicators must be pre-calculated here
        self.ma = self.I(talib.MA, self.data.Close, self.ma_window)
        self.macd, self.macd_signal, self.macd_hist = self.I(
            talib.MACD,
            self.data.Close,
            self.fastperiod,
            self.slowperiod,
            self.signalperiod
        )

    def next(self):
        price = self.data.Close[-1]
        m1 = self.ma[-1]
        
        # Calculate slope (Change over 1 period)
        # Avoid division by zero if ma is 0
        slope_pct = (self.ma[-1] - self.ma[-2]) / self.ma[-2] if self.ma[-2] != 0 else 0

        # ENTRY LOGIC: If no position is open
        if not self.position:
            # Condition: Bullish crossover and upward MA slope
            if slope_pct > 0 and crossover(self.macd, self.macd_signal):
                self.buy()

        # EXIT LOGIC: If a position is already open
        else:
            # Condition 1: Bearish crossover with downward slope
            # Condition 2: Price drops 2% below the MA (Breathing room)
            is_bearish_cross = crossover(self.macd_signal, self.macd)
            
            if (price < m1 * 0.9) or (slope_pct < 0 and is_bearish_cross):
                self.position.close()
```

```python
bt = Backtest(data, MA_MACD, cash=10000)
```

```python
stats = bt.optimize(
    ma_window = range(20,50,2),
    fastperiod = range(10,20,2),
    slowperiod = range(20,30,2),
    signalperiod = range(5,15,1),
    maximize = 'Equity Final [$]',
    max_tries = 200
)
```

```python
stats
```

```python
bt.plot()
```
