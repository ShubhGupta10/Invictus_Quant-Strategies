```python
from backtesting import Backtest, Strategy
from backtesting.lib import crossover
import talib,pandas, numpy
import yfinance as yf
```

```python
gc = yf.Ticker("GC=F")
data = gc.history('max')
```

```python
split_idx = int(len(data)*0.8)

iss_data = data.iloc[:split_idx]
oss_data = data.iloc[split_idx:]
```

```python
oss_data
```

```python
class CCI(Strategy):
    
    cci_period = 20
    ema_period = 200
    vol_period = 20

    def init(self):
        self.ema = self.I(talib.EMA, self.data.Close, self.ema_period)
        self.cci = self.I(talib.CCI, self.data.High, self.data.Low, self.data.Close, self.cci_period)
        self.avg_vol = self.I(talib.SMA, self.data.Volume.astype(float), self.vol_period)

        
    def next(self):
        price = self.data.Close[-1]
        vol = self.data.Close[-1]

        if crossover(self.cci, -100) and price>self.ema and (vol>(1.2*self.avg_vol)):
            self.buy(sl=0.95*price)

        elif price<self.ema:
            self.position.close()
```

```python
bt = Backtest(iss_data, CCI, cash=10_000)
```

```python
stats = bt.optimize(
    ema_period = range(150, 250, 10),
    cci_period = range(10,30,2),
    vol_period = range(10,50,5),
    maximize = "Equity Final [$]"
)
```

```python
bt_opt = Backtest(oss_data, CCI_optimized, cash = 10_000)
```

```python
stats_opt = bt_opt.run()
```
